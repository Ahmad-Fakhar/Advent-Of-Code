**PROGRAM 01 | Cinema Ticket Calculator**

```python
print("PROGRAM 01: Cinema Ticket Calculator")
try:
    age = int(input("Enter customer's age: "))
    ticket_count = int(input("Enter number of tickets: "))

    if age < 0 or ticket_count <= 0:
        print("Please enter a valid age and at least one ticket.")
    else:
        if age >= 12:
            ticket_price = 300
        else:
            ticket_price = 180

        total_bill = ticket_price * ticket_count
        print(f"Price per ticket: Rs. {ticket_price}")
        print(f"Total bill: Rs. {total_bill}")
except ValueError:
    print("Please enter whole numbers for age and ticket count.")
```

**PROGRAM 02 | Name Tag Generator**

```python
print("PROGRAM 02: Name Tag Generator")
first_name = input("Enter first name: ").strip()
last_name = input("Enter last name: ").strip()

if first_name and last_name:
    full_name = f"{first_name.title()} {last_name.title()}"
    initials = f"{first_name[0].upper()}.{last_name[0].upper()}."
    letter_count = len(first_name.replace(" ", "")) + len(last_name.replace(" ", ""))

    print(f"Full name: {full_name}")
    print(f"Initials: {initials}")
    print(f"Total letters: {letter_count}")
else:
    print("Please enter both a first name and a last name.")
```

**PROGRAM 03 | Five-Day Step Tracker**

```python
print("PROGRAM 03: Five-Day Step Tracker")
daily_steps = []

try:
    for day in range(1, 6):
        steps = int(input(f"Enter steps walked on day {day}: "))
        if steps < 0:
            raise ValueError("Steps cannot be negative.")
        daily_steps.append(steps)

    total_steps = sum(daily_steps)
    highest_steps = max(daily_steps)
    average_steps = total_steps / len(daily_steps)
    qualifying_days = sum(1 for steps in daily_steps if steps >= 6000)

    print(f"Total steps: {total_steps}")
    print(f"Highest daily steps: {highest_steps}")
    print(f"Average steps: {average_steps:.2f}")
    print(f"Days with 6,000 steps or more: {qualifying_days}")
except ValueError:
    print("Please enter a valid non-negative whole number for each day's steps.")
```

**PROGRAM 04 | Stationery Price Manager**

```python
print("PROGRAM 04: Stationery Price Manager")
stationery_prices = {
    "pen": 50,
    "notebook": 200,
    "eraser": 20,
}

print("Current prices:")
for item, price in stationery_prices.items():
    print(f"{item.title()}: Rs. {price}")

item_to_update = input("Which item would you like to update? ").strip().lower()
if item_to_update in stationery_prices:
    try:
        new_price = float(input(f"Enter the new price for {item_to_update}: Rs. "))
        if new_price < 0:
            print("Price cannot be negative. Prices were not changed.")
        else:
            stationery_prices.update({item_to_update: new_price})
            print("\nUpdated prices:")
            for item, price in stationery_prices.items():
                print(f"{item.title()}: Rs. {price:.2f}")
            print(f"Sum of all three prices: Rs. {sum(stationery_prices.values()):.2f}")
    except ValueError:
        print("Please enter a valid numeric price.")
else:
    print("Item not found. Please choose pen, notebook, or eraser.")
```

**PROGRAM 05 | Three-Subject Result Checker**

```python
def average_marks(a: int, b: int, c: int) -> float:
    """Return the average of three subject marks."""
    return (a + b + c) / 3
assert average_marks(60, 70, 80) == 70.0


print("PROGRAM 05: Three-Subject Result Checker")
try:
    mark_1 = int(input("Enter marks for subject 1: "))
    mark_2 = int(input("Enter marks for subject 2: "))
    mark_3 = int(input("Enter marks for subject 3: "))

    if any(mark < 0 or mark > 100 for mark in (mark_1, mark_2, mark_3)):
        print("Marks must be between 0 and 100.")
    else:
        average = average_marks(mark_1, mark_2, mark_3)
        print(f"Average marks: {average:.1f}")
        if average >= 50:
            print("Result: Pass")
        else:
            print("Result: Needs improvement")
except ValueError:
    print("Please enter whole-number marks.")
```

**PROGRAM 06 | Visitor Register**

```python
print("PROGRAM 06: Visitor Register")
visitor_one = input("Enter the first visitor's name: ").strip()
visitor_two = input("Enter the second visitor's name: ").strip()

# Write mode creates or replaces the file; append mode adds a new line.
with open("visitors.txt", "w", encoding="utf-8") as visitor_file:
    visitor_file.write(visitor_one + "\n")

with open("visitors.txt", "a", encoding="utf-8") as visitor_file:
    visitor_file.write(visitor_two + "\n")

with open("visitors.txt", "r", encoding="utf-8") as visitor_file:
    visitor_names = visitor_file.read()

print("Visitors saved in visitors.txt:")
print(visitor_names, end="" if visitor_names.endswith("\n") else "\n")
```

**PROGRAM 07 | Movie Watchlist with JSON**

```python
import json
print("PROGRAM 07: Movie Watchlist with JSON")
movies = [
    {"title": "The Martian", "watched": True},
    {"title": "Interstellar", "watched": False},
    {"title": "Hidden Figures", "watched": False},
]

with open("watchlist.json", "w", encoding="utf-8") as json_file:
    json.dump(movies, json_file, indent=4)

with open("watchlist.json", "r", encoding="utf-8") as json_file:
    loaded_movies = json.load(json_file)

print(f"Second movie: {loaded_movies[1]['title']}")
unwatched_titles = [
    movie["title"] for movie in loaded_movies if not movie["watched"]
]
print("Movies not yet watched:")
for title in unwatched_titles:
    print(f"- {title}")
```

**PROGRAM 08 | Shared-Bill Calculator**

```python
print("PROGRAM 08: Shared-Bill Calculator")
try:
    total_bill = float(input("Enter the total restaurant bill (Rs.): "))
    number_of_friends = int(input("Enter the number of friends sharing the bill: "))

    if number_of_friends <= 0:
        print("The number of friends must be greater than zero.")
    elif total_bill < 0:
        print("The bill cannot be negative.")
    else:
        amount_each = total_bill / number_of_friends
        print(f"Each person should pay: Rs. {amount_each:.2f}")
except ValueError:
    print("Invalid input. Enter a numeric bill and a whole number of friends.")
```

**PROGRAM 09 | Parcel Delivery Classes**

```python
class Parcel:
    """Represent a standard parcel delivery."""

    def __init__(self, item_name: str, weight: float):
        self.item_name = item_name
        self.weight = weight

    def delivery_fee(self) -> float:
        """Calculate the standard delivery fee."""
        return self.weight * 50


class ExpressParcel(Parcel):
    """Represent an express parcel with a higher delivery rate."""

    def delivery_fee(self) -> float:
        """Calculate the express delivery fee."""
        return self.weight * 80



print("PROGRAM 09: Parcel Delivery Classes")
normal_parcel = Parcel("Books", 2)
express_parcel = ExpressParcel("Documents", 2)

print(f"Normal parcel ({normal_parcel.item_name}) fee: Rs. {normal_parcel.delivery_fee():.2f}")
print(f"Express parcel ({express_parcel.item_name}) fee: Rs. {express_parcel.delivery_fee():.2f}")
```
